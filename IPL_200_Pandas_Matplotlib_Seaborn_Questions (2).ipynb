### Question 175

### Question 180
Create a multi-plot figure showing season-wise total runs, sixes, fours, and extras using four subplots.

```python
import matplotlib.pyplot as plt

# Merge deliveries with season information
merged = deliveries.merge(
    matches[["id", "season"]],
    left_on="match_id",
    right_on="id"
)

# Season-wise calculations
season_runs = merged.groupby("season")["total_runs"].sum()

season_sixes = (
    merged[merged["batsman_runs"] == 6]
    .groupby("season")
    .size()
)

season_fours = (
    merged[merged["batsman_runs"] == 4]
    .groupby("season")
    .size()
)

season_extras = merged.groupby("season")["extra_runs"].sum()

# Create 4 subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Total runs
axes[0, 0].plot(season_runs.index, season_runs.values, marker="o")
axes[0, 0].set_title("Season-wise Total Runs")
axes[0, 0].set_xlabel("Season")
axes[0, 0].set_ylabel("Total Runs")

# 2. Sixes
axes[0, 1].plot(season_sixes.index, season_sixes.values, marker="o")
axes[0, 1].set_title("Season-wise Total Sixes")
axes[0, 1].set_xlabel("Season")
axes[0, 1].set_ylabel("Sixes")

# 3. Fours
axes[1, 0].plot(season_fours.index, season_fours.values, marker="o")
axes[1, 0].set_title("Season-wise Total Fours")
axes[1, 0].set_xlabel("Season")
axes[1, 0].set_ylabel("Fours")

# 4. Extras
axes[1, 1].plot(season_extras.index, season_extras.values, marker="o")
axes[1, 1].set_title("Season-wise Total Extras")
axes[1, 1].set_xlabel("Season")
axes[1, 1].set_ylabel("Extras")

plt.tight_layout()
plt.show()
```

### Question 200
Create one figure with two Seaborn plots side by side: top 10 run scorers and top 10 six hitters.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Top 10 run scorers
top_10_runs = (
    deliveries.groupby("batsman")["batsman_runs"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

# Top 10 six hitters
top_10_sixes = (
    deliveries[deliveries["batsman_runs"] == 6]
    .groupby("batsman")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .reset_index(name="sixes")
)

# One figure with two plots side by side
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Top 10 run scorers
sns.barplot(
    data=top_10_runs,
    x="batsman_runs",
    y="batsman",
    ax=axes[0]
)
axes[0].set_title("Top 10 Run Scorers")
axes[0].set_xlabel("Total Runs")
axes[0].set_ylabel("Batsman")

# Plot 2: Top 10 six hitters
sns.barplot(
    data=top_10_sixes,
    x="sixes",
    y="batsman",
    ax=axes[1]
)
axes[1].set_title("Top 10 Six Hitters")
axes[1].set_xlabel("Number of Sixes")
axes[1].set_ylabel("Batsman")

plt.tight_layout()
plt.show()
```
